```python
balance = 1000
interest_rate = 0.05

for year in range(1, 6):
    balance = balance * (1 + interest_rate)
    print(f"Year {year}: ${balance:.2f}")
```

#### 9. Write the code necessary to calculate the sum of the square differences of a group of values : 
58, 32, 37, 41, 36, 36, 54, 37, 25, 53
##### $\Sigma $(${x}$ - $\bar{x}$)$^2$

```python
values = 58, 32, 37, 41, 36, 36, 54, 37, 25, 53

average = sum(values) / len(values)
squared_differences = []

for value in values:
    difference = value - average
    squared_differences.append(difference ** 2)

result = sum(squared_differences)

print("Average:", average)
print("Sum of Squared Differences:", round(result, 2))
```

#### 11. Create a function the will simulate rolling two die and return the value of each and and the sum 

```python
import random

def roll_two_dice():
    die_one = random.randint(1,6)
    die_two = random.randint(1,6)
    total = die_one + die_two

    return die_one, die_two, total

first, second, total = roll_two_dice()

print("First die:", first)
print("Second die:", second)
print("Total:", total)
```

#### 12. Write the code needed to evaluate whether a value in a group of values is negative or postive and prints an approriate message, i.e. postive, negative 
-1.59, 2.36, 1.69, 1.13, -0.91, 1.48, -0.34, 1.31, -0.74, 0.2

```python
values = [-1.59, 2.36, 1.69, 1.13, -0.91, 1.48, -0.34, 1.31, -0.74, 0.2]

for value in values:
    if value > 0:
        print(value, "is positive")
    elif value < 0:
        print(value, "is negative")
    else:
        print(value, "is zero")
```

#### 14.  Complete the following:
1. Write a program that generates 500 random integers from 1 to n, for example, the first integer will be 1 the second will be 1 or 2, the fifth between 1 and 5, and so on.
2. The program should write each integer to a file 
3. Once all integers are written close the file

##### Part 2
1. Open and read the file into a list
2. Create a line plot of the data

```python
import random

with open ("random_integers.txt", "w") as file:
    for n in range(1, 501):
        file.write(str(random.randint(1, n)) + "\n")
```

```python
numbers = []

with open("random_integers.txt", "r") as file:
    for line in file:
        number = int(line.strip())
        numbers.append(number)

print("The numbers read:", len(numbers))
print("First 10:", numbers[:10])
```
